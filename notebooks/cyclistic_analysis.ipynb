## Step 1: Environment Setup
In this step, we will import the required libraries (`pandas`, `numpy`, `matplotlib`, `seaborn`, `os`) to perform data manipulation, numerical calculations, and visualizations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

print("Libraries successfully imported!")
```

## Step 2: Load and Merge Data
We have 12 separate monthly CSV files containing millions of trip records. We will load each file and concatenate them into a single pandas DataFrame.

```python
# Define directory containing data
DATA_DIR = '../data'

# List all CSV files in the folder
csv_files = [f for f in os.listdir(DATA_DIR) if f.endswith('.csv')]
print(f"Found {len(csv_files)} CSV files in data folder:")
for csv in sorted(csv_files):
    print(f" - {csv}")

# Load and concatenate all CSV files
print("\nLoading and merging data (this may take 1-2 minutes)...")
dfs = [pd.read_csv(os.path.join(DATA_DIR, f)) for f in csv_files]
df = pd.concat(dfs, ignore_index=True)

print(f"\nData successfully merged! Total rows: {df.shape[0]:,}, Total columns: {df.shape[1]}")
```
